1. Load the dataset into dataframe.
2. perform a "Left Join" on the TransactionID column to attach the train_identity data to the train_transaction data.  

```python
import pandas as pd


transaction_path = '../raw_dataset/train_transaction.csv'
identity_path = '../raw_dataset/train_identity.csv'

print("Loading datasets ...")
train_transaction = pd.read_csv(transaction_path)
train_identity = pd.read_csv(identity_path)

print("Merging datasets on TransactionID...")
train = pd.merge(train_transaction, train_identity, on='TransactionID', how='left')

# Print the dimensions of our new combined dataset
print(f"Merged dataset shape: {train.shape}")

# Display the first 5 rows to verify it worked
train.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from matplotlib.backends.backend_pdf import PdfPages
from scratch_implementation.custom_imputer import ScratchImputer

print("Analyzing all 434 columns and applying imputation rules... (This may take 1-3 minutes)")

# 1. Initialize and run the imputer on the FULL dataset
imputer = ScratchImputer(missing_threshold=0.70, skew_threshold=1.0)
train_clean, strategy_report = imputer.fit_transform(train)

# 2. Before and After Details
print("\n--- BEFORE VS AFTER MISSING VALUE TREATMENT ---")
print(f"BEFORE:")
print(f"  Shape: {train.shape}")
print(f"  Total Missing Cells: {train.isnull().sum().sum():,}")
print(f"\nAFTER:")
print(f"  Shape: {train_clean.shape}")
print(f"  Total Missing Cells: {train_clean.isnull().sum().sum():,}")

# --- FIX: Proper Directory Routing and PDF Generation ---

# 3. Save Tabular Summary as PDF
tabular_dir = 'summary/tabular_statistics/'
os.makedirs(tabular_dir, exist_ok=True)
pdf_path = os.path.join(tabular_dir, 'imputation_reasoning_report.pdf')

print("\nGenerating paginated PDF report...")
with PdfPages(pdf_path) as pdf:
    rows_per_page = 35 # Slightly fewer rows to ensure the 'Reason' text fits
    chunks = [strategy_report[i:i + rows_per_page] for i in range(0, len(strategy_report), rows_per_page)]
    
    for idx, chunk in enumerate(chunks):
        fig, ax = plt.subplots(figsize=(14, len(chunk) * 0.4 + 1)) # Wider figure for text
        ax.axis('tight')
        ax.axis('off')
        
        table = ax.table(cellText=chunk.values, colLabels=chunk.columns, loc='center', cellLoc='left')
        table.auto_set_font_size(False)
        table.set_fontsize(9)
        table.scale(1.2, 1.5)
        
        # Style the header row
        for (row, col), cell in table.get_celld().items():
            if row == 0:
                cell.set_text_props(weight='bold', color='white')
                cell.set_facecolor('#4c72b0')
            elif col == 2: # Give the 'Reason' column more breathing room
                cell.set_width(0.6) 
                
        plt.title(f"Imputation Reasoning Report (Page {idx + 1} of {len(chunks)})", weight='bold', pad=20)
        pdf.savefig(fig, bbox_inches='tight')
        plt.close()

print(f"✅ Full tabular reasoning report saved to: {pdf_path}")

# 4. Save Pie Chart to Visualization Folder
vis_dir = 'summary/visualization/'
os.makedirs(vis_dir, exist_ok=True)
pie_chart_path = os.path.join(vis_dir, 'imputation_strategies_pie_chart.png')

plt.figure(figsize=(9, 9))
strategy_counts = strategy_report['Strategy'].value_counts()

colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
plt.pie(strategy_counts, labels=strategy_counts.index, autopct='%1.1f%%', 
        startangle=140, colors=colors[:len(strategy_counts)], 
        wedgeprops={'edgecolor': 'black', 'linewidth': 1})

plt.title('Distribution of Imputation Strategies Across All Features', weight='bold', size=14)

plt.savefig(pie_chart_path, bbox_inches='tight', dpi=300)
plt.close()
print(f"✅ Pie chart visualization saved to: {pie_chart_path}\n")
```
